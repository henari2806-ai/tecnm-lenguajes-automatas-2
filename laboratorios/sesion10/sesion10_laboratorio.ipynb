````python
%%writefile artesano_sesion10_c.c
/* artesano_sesion10_c.c - Fase 1: El Artesano (sin IA)
 * Mini-evaluador de expresiones (Hito 1).
 *
 * Gramatica que use:
 *   expr     -> termino (('+' | '-') termino)*
 *   termino  -> primaria (('*' | '/') primaria)*
 *   primaria -> numero | '(' expr ')'
 */
#include "api_sesion10.h"
#include <stdio.h>
#include <string.h>

#include <ctype.h>

typedef struct { const char *s; int error; } Parser;

static void saltar(Parser *p) {
    while (*p->s == ' ') p->s++;
}

long expr(Parser *p);

long primaria(Parser *p) {
    saltar(p);

    /* caso 1: parentesis, resuelvo lo de adentro y espero el ')' */
    if (*p->s == '(') {
        p->s++;
        long v = expr(p);
        saltar(p);
        if (p->error != 0) return 0;
        if (*p->s != ')') {   /* se abrio y nunca se cerro */
            p->error = 2;
            return 0;
        }
        p->s++;
        return v;
    }

    /* caso 2: un numero, lo voy armando digito por digito */
    if (isdigit((unsigned char)*p->s)) {
        long v = 0;
        while (isdigit((unsigned char)*p->s)) {
            v = v * 10 + (*p->s - '0');
            p->s++;
        }
        return v;
    }

    /* caso 3: no es ni numero ni parentesis, sintaxis invalida */
    p->error = 2;
    return 0;
}

long termino(Parser *p) {
    long v = primaria(p);
    saltar(p);
    while (p->error == 0 && (*p->s == '*' || *p->s == '/')) {
        char op = *p->s;
        p->s++;
        long d = primaria(p);
        if (p->error != 0) return 0;
        if (op == '*') {
            v = v * d;
        } else {
            if (d == 0) {   /* reviso antes de dividir */
                p->error = 1;
                return 0;
            }
            v = v / d;
        }
        saltar(p);
    }
    return v;
}

long expr(Parser *p) {
    long v = termino(p);
    saltar(p);
    while (p->error == 0 && (*p->s == '+' || *p->s == '-')) {
        char op = *p->s;
        p->s++;
        long d = termino(p);
        if (p->error != 0) return 0;
        if (op == '+') {
            v = v + d;
        } else {
            v = v - d;
        }
        saltar(p);
    }
    return v;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int sin_error;
    if (error == NULL) error = &sin_error;
    *error = 0;
    if (fuente == NULL) { *error = 2; return 0; }
    Parser p = { fuente, 0 };
    long v = expr(&p);
    saltar(&p);
    if (p.error == 0 && *p.s != '\0') p.error = 2;
    *error = p.error;
    return p.error != 0 ? 0 : v;
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle

CONTEXTO: Estoy en la materia Lenguajes y Autómatas II, programando en C (estándar C11).
Estoy haciendo el Hito 1: un mini-evaluador que lee una expresión aritmética desde texto.
La interfaz está en api_sesion10.h. Se compila con gcc -std=c11 -Wall -Wextra -Werror, o
sea que cualquier advertencia es error.

ROL: Actúa como ingeniero de software de sistemas y compiladores, con énfasis en
analizadores sintácticos descendentes recursivos y manejo de errores.

ESPECIFICACIÓN: Implementa exactamente esta firma, sin cambiarla:
  long evaluar_expresion(const char *fuente, int *error);
Reglas: lee enteros, los operadores + - * / y paréntesis, ignorando espacios; sin variables
ni menos unario; respeta la precedencia (* y / antes que + y -) y asocia a la izquierda,
así que 10-2-3 da 5; *error vale 0 si todo salió bien, 1 si hay división entre cero y 2 si
la sintaxis es inválida (por ejemplo "2+", "(2+3" o texto sobrante); comprueba el divisor
antes de dividir; no debe tener main().

OUTPUT: Entrégame solo el archivo .c con #include "api_sesion10.h", con comentarios breves
y el análisis de complejidad de la función.

````python
%%writefile ia_sesion10_c.c
/* ia_sesion10_c.c - Fase 2: El Cientifico */
#include "api_sesion10.h"
#include <ctype.h>

typedef struct { const char *s; int error; } Lector;

static void espacios(Lector *l) {
    while (isspace((unsigned char)*l->s)) {
        l->s++;
    }
}

static long expresion(Lector *l);

static long primaria(Lector *l) {
    espacios(l);
    if (*l->s == '(') {
        l->s++;
        long v = expresion(l);
        espacios(l);
        if (l->error) {
            return 0;
        }
        if (*l->s != ')') {
            l->error = 2;
            return 0;
        }
        l->s++;
        return v;
    }
    if (!isdigit((unsigned char)*l->s)) {
        l->error = 2;
        return 0;
    }
    long v = 0;
    while (isdigit((unsigned char)*l->s)) {
        v = v * 10 + (*l->s - '0');
        l->s++;
    }
    return v;
}

static long termino(Lector *l) {
    long v = primaria(l);
    for (;;) {
        espacios(l);
        char op = *l->s;
        if (l->error || (op != '*' && op != '/')) {
            return v;
        }
        l->s++;
        long d = primaria(l);
        if (l->error) {
            return 0;
        }
        if (op == '*') {
            v *= d;
        } else if (d == 0) {
            l->error = 1;
            return 0;
        } else {
            v /= d;
        }
    }
}

static long expresion(Lector *l) {
    long v = termino(l);
    for (;;) {
        espacios(l);
        char op = *l->s;
        if (l->error || (op != '+' && op != '-')) {
            return v;
        }
        l->s++;
        long d = termino(l);
        if (l->error) {
            return 0;
        }
        v = (op == '+') ? v + d : v - d;
    }
}

long evaluar_expresion(const char *fuente, int *error) {
    if (fuente == NULL) {
        *error = 2;
        return 0;
    }
    Lector l = { fuente, 0 };
    long v = expresion(&l);
    espacios(&l);
    if (l.error == 0 && *l.s != '\0') {
        l.error = 2;
    }
    *error = l.error;
    return l.error ? 0 : v;
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("5-10",&e); (void)e; verificar_int("\"5-10\"", (long)(-5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
